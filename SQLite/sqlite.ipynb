## SQL and SQLite

SQL (Structured Query language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless and zero configuration database engine that is widely used for embedded database systems.

```python
import sqlite3
```

```python
## Connect to an SQLite database
connection = sqlite3.connect('example.db')
connection
```

```python
cursor =connection.cursor()
```

```python
## Create a Table
cursor.execute('''
Create Table If Not Exists employees(
                    id Integer Primary key,
                    name Text Not Null,
                    age Integer,
                    department text

                )
'''

)

## Commit the Changes 
connection.commit()
```

```python
cursor.execute('''
Select * from employees

                
'''

)
```

```python
## Insert the Data in sqlite table
cursor.execute('''
    Insert Into employees(name,age,department)
                    values("Soham",22,"Computer Science")
    '''
)
cursor.execute('''
    Insert Into employees(name,age,department)
                    values("Bob",32,"AI")
    '''
)
cursor.execute('''
    Insert Into employees(name,age,department)
                    values("Ron",25,"Data Science")
    '''
)
cursor.execute('''
    Insert Into employees(name,age,department)
                    values("Mike",25,"law")
    '''
)
cursor.execute('''
    Insert Into employees(name,age,department)
                    values("Jonathan",32,"Finance")
    '''
)

connection.commit()
```

```python
## Query the data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## Update the data in the table
cursor.execute(
    '''
UPDATE employees
Set age=34
where name="Bob"
'''
)

connection.commit()
```
