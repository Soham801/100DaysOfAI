Inheritance

```python
## Parent class

class Car:
    # Constructor
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The person will drive a {self.enginetype} car.")
```

```python
car1 = Car(4,4,"Petrol")
car1.drive()
```

```python
# Inheritance
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_selfdrive):
        super().__init__(windows,doors,enginetype)
        self.is_selfdrive=is_selfdrive,
    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_selfdrive}")
```

```python
tesla1 = Tesla(4,2,"Electric",True)
tesla1.selfdriving()
```

```python
class Mustang(Car):
    def __init__(self,windows , doors,enginetype,is_powerful):
        super().__init__(windows,doors,enginetype),
        self.is_powerful=is_powerful

    def drive(self):
        print(f"Mustang runs on a {self.enginetype} based engine and The Car is Powerful - {self.is_powerful}")
        
```

```python
mustang1 = Mustang(2,2,"Gasoline",True)
mustang1.drive()
```

```python
### Multiple Inheritance 
## When a class inherits from more than one base class.

## Base class 1

class Animal:
    def __init__(self,name:str):
        self.name=name

        def speak(self):
            print("Subclass must implement this method")

## Base Class 2

class Pet:
    def __init__(self , owner:str):
        self.owner=owner
```

```python
### Derives class

class Dog(Animal,Pet):
    def __init__(self, name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)

    def barks(self):
        return(f"This golden retriver is named as {self.name} by {self.owner}")
```

```python
### Calling multiple inherited Classes

dog = Dog("Leo","Soham")
dog.barks()
```
