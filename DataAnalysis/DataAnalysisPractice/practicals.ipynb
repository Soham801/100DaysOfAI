```python
## Practical 2 - Create a 1D array of numbers from 1 to 20. Reshape it into a 4x5 matrix (2D array) and print the shape and data type. 

array1 =  np.arange(1,21)
print(f"The Created 1D Array - {array1}")


array2 = array1.reshape(4,5)
print(f"Updated 1D Array into a 4x5 Array - {array2}")

array2.dtype
```

```python
## Practical 3 - Create an array of 5 product prices and perform mathematical operations on entire datasets at once. 

prices = np.array([222,344,199.5,62.2,205])


print(f"Original Prices - {prices}")


print(f"Lowest value of a item - {np.min(prices)}")
print(f"Highest value of a item - {np.max(prices)}")
print(f"Average value of items in prices - {np.mean(prices)}")
print(f"Sum of all values of items in prices - {np.sum(prices)}")

discounted_prices = prices * 0.80
print(f"Discount on items in prices - {discounted_prices}")

gst_on_prices = prices * 1.18
print(f"GST on items in prices - {gst_on_prices}")
```
