Inheritance

```python
## Parent class

class Car:
    # Constructor
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"The person will drive a {self.enginetype} car.")
```

```python
car1 = Car(4,4,"Petrol")
car1.drive()
```

```python
# Inheritance
class Tesla(Car):
    def __init__(self,windows,doors,enginetype,is_selfdrive):
        super().__init__(windows,doors,enginetype)
        self.is_selfdrive=is_selfdrive,
    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_selfdrive}")
```

```python
tesla1 = Tesla(4,2,"Electric",True)
tesla1.selfdriving()
```

```python
class Mustang(Car):
    def __init__(self,windows , doors,engine_type):
        super().__init__(windows,doors,engine_type)
        print(f"Mustang runs on a {engine_type} based engine.")
```
